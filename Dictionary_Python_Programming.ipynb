## 2. Access a value from a dictionary.

```python
# Question: Given the dictionary {"name":"Vikas","Age":18},print the age
person = {"name":"Vikas","Age":18}
print(person["Age"])
```

## 3. Add a new key value pair to a dictionary.

```python
# Question: Add a key "City" with value "New York" to the dictionary person.
person = {"name":"Vikas","Age":18}
person["City"] = "New York"
print(person)
```

## 4. UPdating an existing value in a dictionary. 

```python
# Question: Change the age of "Vikas" to 18 in the given dictionary.
person = {"Name":"Vikas","Age":18}
person["Age"] = 20
print(person)
```

## 5. Delete a key from a dictionary.

```python
# Question: Remove the "Age" key from the dictionary.
person = {"Name":"Vikas","Age":18}
del person["Age"]
print(person)
```

## 7. Use get () method to access value safely.

```python
# Question: Access the "City" key using get (), provide default a "unknow" if not present.
person = {"Name":"Vikas","Age":18}
city = person.get("city","Unknow")
print(city)
```
