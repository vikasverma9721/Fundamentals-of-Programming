```python
# Ques_9  Find the largest of three numbers.
num1 = int(input("Enter the first number"))
num2 = int(input("Enter the second number"))
num3 = int(input("Enter the third number"))
if num1>num2 and num1>num3:
    print(f' {num1} is greatest')
elif num2>num3 and num2>num1:
    print(f' {num2} is greatest')
else:
    print(f' {num3} is greatest')
```

```python
# Ques_10  Check if a letter is a vowel or consonant.
Letter = input("Write a Letter")
vowel = ["A","E","I","O","U"]
if Letter in vowel:
    print(f'{Letter}: is vowel')
else:
    print(f'{Letter}: is consonant')
```

```python
# Ques_11 Check if a number is positive, negative, or zero.
num = int(input("Enter a number"))
if num > 0:
    print(f' {num}: is possitive')
elif num < 0:
    print(f' {num}: is negative')
else:
    print(f' {num}: is zero')
```

```python
# Ques Simple Calculator: Perform +, -, *, / operations.
num1 = float(input("Enter the first number"))
num2 = float(input("Enter the second number"))
operation = input("Write the operation")
a = num1 + num2
b = num1 - num2
c = num1 * num2
d = num1 / num2
if operation =='+':
    result = (f'{num1} + {num2} = {a}')
elif operation =='-':
    result = (f'{num1} - {num2} = {b}')
elif operation =='*':
    result = (f'{num1} * {num2} = {c}')
elif operation == '/':
    result = (f'{num1} / {num2} = {d}')
print(result)
```
